## Численное дифференцирование

Пусть требуется найти значение производной заданной функции $f(x)$ в точке $x = x_0$. Будем полагать, что в нашем распоряжении имеется алгоритм, позволяющий получать значения функции $f(x)$ в некотором конечном интервале аргументов $x$ в окрестности точки $x_0$.  

Вычислим значения данной функции $f(x)$ в равноотстоящих точках:

$$
x_i = x_0 + ih, \quad i = 0, \pm 1, \pm 2, \pm 3, \dots. 
$$

Величину $h$ выберем малой. Тогда точки $x_1 = x_0 + h$ и $x_{−1} = x_0 − h$ можно полагать находящимися в окрестности точки $x_0$. Запишем следующие разложения в ряд Тейлора для функции $f(x)$:  

$$
f(x_1) = f(x_0) + hf'(x_0)+\frac{h^2}{2}f''(x_0) + \frac{h^3}{3!} f'''(x0) + \dots, 
$$
$$
f(x_{-1}) = f(x_0) - hf'(x_0) + \frac{h^2}{2}f''(x_0) - \frac{h^3}{3!} f'''(x0) + \dots
$$

Сначала в разложениях отбросим все члены 3-го порядка и выше. Из-за малости величины h такое приближение вполне допустимо. После этого вычтем одно выражение из другого и получим следующее уравнение:  
$$
f (x_1) − f (x_{−1}) = 2 h f'(x_0). 
$$
Отсюда сразу получается формула для приближенного вычисления производной в точке $x = x_0$: 
$$
f'(x_0) \approx \frac{f(x_0 + h) - f(x_0 - h)}{2h}.
$$

Эта формула называется 3-точечной оценкой первой производной функции $f(x)$. Значение производной может быть уточнено с помощью 5-точечной оценки:

$$
f'(x_0) \approx \frac{f(x_0 -2h) -8 f(x_0-h) + 8 f(x_0+h) - f (x_0 + 2h)}{12h}
$$

Для второй производной функции $f(x)$ в точке $x = x_0$ могут использоваться аналогичные 3- и 5-точечные оценки:

$$
f''(x_0) \approx \frac{f(x_0 + h) - 2f(x_0) + f(x_0-h)}{h^2}
$$

$$
f''(x_0) \approx \frac{-f(x_0 -2h) + 16 f(x_0-h) - 30 f(x_0)+ 16f(x_0 + h) - f(x_0 +2h)}{12h^2}
$$
