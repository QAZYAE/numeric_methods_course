$$
I(f, a, b) = \int_a^b f(x) dx
$$
### Метод трапеций
Разобьем отрезок $[a, b]$ на $N$ одинаковых поддиапазонов шириной $h = \frac{b - a}{N}$. Тогда искомый интеграл представится суммой интегралов функции на каждом из поддиапазонов:

$$
I = \sum_{i=0}^N I_i, \quad I_i = \int_{a+ih}^{a+(i+1)h} f(x)dx.
$$

Аппроксимируем кривую на каждом интервале линейной функцией. Тогда площадь под кривой на каждом интервале можно вычислить как площадь трапеции:

$$
I_i \approx \frac{y_i + y_{i+1}}{2}h
$$
В итоге получится следующая формула для вычисления интеграла:

$$
I \approx h \left( \sum_{i=1}^{N-1} y_k + \frac{y_0 + y_N}{2} \right)
$$

$$
I_0 = h \left( \frac16 y_0 + \frac23 y_1 + \frac16 y2 \right)
$$

$$
I(f, a, b) \approx \frac{h}{3} \left( y_0 + 4S_0 + 2S_E + y_{2M} \right)
$$

$$
S_0 = y_1 + y_3 + ... + y_{2M-1}, \quad S_E = y_2 + y_4 + ... + y_{2M-2}
$$
